```python
#SUM of element in list

my_list = [1,2,3,4,5,6,7,8]

sum =0

for i in my_list:
    sum+=i
print(sum)
```

```python
#MUL of element in list

my_list = [1,2,3,4,5,6,7,8]

mul =1

for i in my_list:
    mul*=i
print(mul)
```

```python
#to find largest element in list

my_list = [1,2,3,699,5,6,7,8]

max = 0

for my_list in my_list:
    if my_list > max:
        max = my_list
print(max)
```

```python
#to find a count and total occurrences of a specific element in list

my_list = [1,2,3,4,5,6,4,7,8]

element = 4
count = 0
total = 0

for i in my_list:
    if i == element:
        count += 1
    total += 1

print("Count of", element, "is:", count)
```
